```python
#W5A16: Viết hàm kiểm tra một số có phải số Amstronhg hay không? Hàm trả về kiểu bool, có tham số là một số nguyên x
def is_amstrong(x):
    if x<0:
        return False
    s = str(x)
    so_chu_so = len(s)
    tong = 0
    for ch in s:
        so = int(ch)
        tong += so**so_chu_so
    return tong == x
```

```python
#W5A17: Viết hàm kiểm tra một số có phải số chính phương hay không? Hàm trả về kiểu bool, có tham số là một số nguyên x
def is_chinhphuong(x):
    if x<0:
        return False
    for i in range(int(x**0.5) +1):
        if i*i == x:
            return True
    return False
```

```python
#W5A18: VIết hàm kiểm tra một số có phải số đối xứng không, xử lý trên dữ liệu kiểu số nguyên.
def is_doixung(x):
    if x<0:
        return False
    temp = x
    doixung = 0
    while temp!=0:
        chusocuoi = temp%10
        doixung = doixung*10 + chusocuoi
        temp //= 10
    if doixung == x:
        return True
    return False
```

```python
#W5A19: VIết hàm kiểm tra một số có phải số kỳ diệu hay không, xử lý trên dữ liệu kiểu số nguyên.
def is_magic_number(n):
    if n <= 0:
        return False
    tong = 0
    temp = n
    while temp>0:
        chuso = temp %10
        tong += chuso
        temp //= 10
    if n%tong == 0:
        return True
    return False
```

```python
#W5A20: Viết hàm detect number type(s) trả về 'int' nếu s là số nguyên, 'float' nếu là số thực, ngược lại 'neither'.
def detect_number_type(s):
  s = s.strip()
  if s == "":
    return "neither"
  if s[0] in "+-":
    s = s[1:]
  if s == "":
    return "neither"
  dot_count = 0
  digit_count = 0
  for ch in s:
    if ch == ".":
      dot_count += 1
      if dot_count > 1:
        return "neither"
    elif "0" <= ch <= "9":
      digit_count += 1
    else:
      return "neither"

  if digit_count == 0:
    return "neither"
  if dot_count == 1:
    return "float"
  return "int"
```

```python
#W5A24: Viết hàm are collinear (x1,y1,x2,y2,x3,y3) kiểm tra ba điểm có thẳng hàng hay không.
def are_collinear(x1,y1,x2,y2,x3,y3):
  return (x2 - x1) * (y3 - y1) == (y2 - y1) * (x3 - x1)
```

```python
#W5A25: Viết hàm point_in_quadrilateral(quad, p) kiểm tra điểm p có năm trong hoặc trên biên của tứ giác quad hay không. Trong đó: quad là một tứ giác (4 đỉnh – 4 điểm), p là một điểm cần kiểm tra. VD: quad =[(0,0), (4,0), (4,3), (0,3)], p=(2,1)
def point_in_quadrilateral(quad, p):
    (x1, y1), (x2, y2), (x3, y3), (x4, y4) = quad
    px, py = p
    s_quad = (abs(x1 * (y2 - y3) + x2 * (y3 - y1) + x3 * (y1 - y2)) + abs(x1 * (y3 - y4) + x3 * (y4 - y1) + x4 * (y1 - y3))) / 2
    s_sum = (abs(px * (y1 - y2) + x1 * (y2 - py) + x2 * (py - y1))+ abs(px * (y2 - y3) + x2 * (y3 - py) + x3 * (py - y2))+ abs(px * (y3 - y4) + x3 * (y4 - py) + x4 * (py - y3))+ abs(px * (y4 - y1) + x4 * (y1 - py) + x1 * (py - y4))) / 2
    return s_quad == s_sum
```

```python
#W5A27: Viết hàm trả về số lượng chữ số khác 0 trong n số nguyên nhập vào từ bàn phím (một list)
def countkhac0(n):
    cnt = 0
    for x in n:
        for ch in str(x):
            if ch not in ("0", "-"):
                cnt += 1
    return cnt  
```

```python
#W5A28: Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3).
import math
def tinhgoc(A,B,C):
    x1,y1 = A
    x2,y2 = B
    x3,y3 = C
    ux,uy = x1-x2,y1-y2
    vx,vy = x3-x2, y3 - y2
    dot = ux*vx + uy*vy
    lenu = (ux**2 + uy**2)**0.5
    leny = (vx**2 + vy**2)**0.5
    cosb = (dot)/(lenu*leny)
    return math.degrees(math.acos(cosb))
```

```python
#W5A29: Viết hàm trả về giá trị lớn nhất và nhỏ nhất của ba số a,b,c dưới dạng list [max, min].
def maxmin(a,b,c):
    lonnhat = a
    nhonhat = a
    if b>lonnhat:
        lonnhat = b
    if b <nhonhat:
        nhonhat = b
    if c>lonnhat:
        lonnhat = c
    if c<nhonhat:
        nhonhat = c
    return [lonnhat,nhonhat]
```

```python
#W5A30: Viết hàm tính hệ số nhị thức C(n,k). 
def C(n,k):
    if k < 0 or k > n:
        return 0
    ngiaithua = 1
    kgiaithua = 1
    ntrukgiaithua = 1
    for i in range(1,n+1):
        ngiaithua *= i
    for i in range(1,k+1):
        kgiaithua *= i
    for i in range(1,n-k+1):
        ntrukgiaithua *= i
    ans = ngiaithua // (kgiaithua*ntrukgiaithua)
    return ans
```

```python
#W5A31: Viết hàm tính tổng bình phương các số từ 1 đến n.
def tongbinhphuong(n):
    if n<1:
        return False
    tong = 0
    for i in range(1,n+1):
        tong = tong + i**2
    return tong
```

```python
#W5A33: Viết hàm smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính tổng tiền taxi theo quy tắc giá mở cửa, giá km, phí chờ, phụ phí ban đêm và giảm giá.
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
  if km <= 0 and wait_minutes <= 0:
    return 0
  if car_type == 4:
    base_fare = 12000
    rate_mid = 14000
    rate_far = 11000
  elif car_type == 7:
    base_fare = 15000
    rate_mid = 16000
    rate_far = 13000
  else:
    raise ValueError("car_type phải là 4 hoặc 7")
  if km <= 0:
    distance_fare = 0
  elif km <= 1:
    distance_fare = base_fare
  elif km <= 20:
    distance_fare = base_fare + (km - 1) * rate_mid
  else:
    distance_fare = base_fare + 19 * rate_mid + (km - 20) * rate_far
  free_wait = 5
  wait_rate = 500
  wait_fare = max(0, wait_minutes - free_wait) * wait_rate
  total = distance_fare + wait_fare
  is_night = start_hour >= 22 or start_hour < 5
  if is_night:
    total *= 1.2
  if km > 30 and not is_night:
    total *= 0.9

  return round(total)


print(smart_taxi_fare(4, 10, 5, 6))
print(smart_taxi_fare(7, 23, 12, 0))
```

```python
#W5A34: Viết hàm sum_1_to_n(n) trả về tổng các số x từ 1 đến n, nếu x có các chữ số đôi một khác nhau. Nếu n<=0 trả về 0.
def sum_1_to_n(n):
    if n<=0:
        return 0
    tong = 0
    for x in range(1,n+1):
        temp = x
        dagap =[]
        hople = True
        while temp > 0:
            chuso = temp % 10
            if chuso in dagap:
                hople = False
                break
            dagap.append(chuso)
            temp //= 10
        if hople:
            tong += x
    return tong

print(sum_1_to_n(30))
```
