```python
#Bài 3 : Tính tam giác: Nhập vào ba số a,b,c bất kì. Kiểm tra xem nó có thể là độ dài ba cạnh hay không, nếu không thì in ra màn hình 'Khong la ba canh cua mot tam giac'. Ngược lại, kiểm tra xem nó là tam giác gì: Đều, Cân, Vuông, hay Tam giác thường. Tính diện tích, chu vi của tam giác và in ra màn hình.
a,b,c = map(float,input().split())
if a > 0 and b >0 and c > 0 and a+b > c and a+c > b and b+c>a:
    if a==b==c:
        print("Tam giác đều")
    elif a==b or a==c or b==c:
        print("Tam giác cân")
    if a**2 + b**2 == c**2 or a**2 + c**2 == b**2 or b**2 + c**2 == a**2:
        print("Tam giác vuông")
    else:
        print("Tam giác thường")
    import math
    p = (a+b+c)/2
    cv = a+b+c
    dt = math.sqrt(p*(p-a)*(p-b)*(p-c))
    print(cv,dt)
else:
    print("Khong la ba canh cua mot tam giac")
```

```python
#Bài 5: Viết chương trình giải phương trình bậc 2 (ax^2+bx+c=0)
import math
a,b,c= map(float,input().split())
if a == 0:
    if b == 0:
        if c == 0:
            print("Vô số nghiệm")
        else:
            print("Vô nghiệm")
    else:
        x = -c / b
        print(f"x = {x}")
else:
    delta = b**2 - 4 * a * c

    if delta < 0:
        print("Vô nghiệm")
    elif delta == 0:
        x = -b / (2 * a)
        print(f"x = {x}")
    else:
        x1 = (-b + math.sqrt(delta)) / (2 * a)
        x2 = (-b - math.sqrt(delta)) / (2 * a)
        print(f"x1 = {x1}")
        print(f"x2 = {x2}")
```

```python
#Bài 8: Giải hệ phương trình tuyến tính:
#ax+by=m
#cx+dy=n
a,b,m,c,d,n = map(float,input().split())
D = a * d - b * c
Dx = m * d - b * n
Dy = a * n - c * m
if D != 0:
    x = Dx / D
    y = Dy / D
    print(f"x = {x}, y = {y}")
else:
    if Dx == 0 and Dy == 0:
        print("Vô số nghiệm")
    else:
        print("Vô nghiệm")
```

```python
#Bài 10: Nhập vào tâm và bán kính của một đường tròn. Sau đó nhập vào một điểm A(x,y) bất kì và kiểm tra xem nó có thuộc đường tròn hay không?
x0, y0 = map(float, input().split())
R = float(input())
if R < 0:
    print("ERROR")
else:
    x, y = map(float, input().split())
    dbinh= (x - x0) ** 2 + (y - y0) ** 2
    Rbinh = R ** 2
    if dbinh == Rbinh:
        print(f"Điểm A({x}, {y}) nằm trên đường tròn")
    elif dbinh < Rbinh:
        print(f"Điểm A({x}, {y}) nằm trong đường tròn")
    else:
        print(f"Điểm A({x}, {y}) nằm ngoài đường tròn")
```

```python
#Bài 10 cách 2
x0, y0 = map(float, input().split())
R = float(input())
if R >= 0:
    x, y = map(float, input().split())
    d2 = (x - x0)**2 + (y - y0)**2
    if d2 == R**2: print("Điểm nằm trên đường tròn")
    elif d2 < R**2: print("Điểm nằm trong đường tròn")
    else: print("Điểm nằm ngoài đường tròn")
else:
    print("ERROR")
```
