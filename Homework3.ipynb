```python
#Bài 4: Viết chương trình in ra hình tam giác cân có độ cao h được nhập từ bàn phím
h = int(input())
for i in range(1, h + 1):
    spaces = " " * (h - i)
    stars = "*" * (2 * i - 1)
    print(spaces + stars)
```

```python
#Bài 8: Nhập vào số nguyên dương a và b, in toàn bộ ước dương chung của a và b
a,b = map(int,input().split())
be = min(a,b)
for i in range(1,be+1):
    if a%i == 0 and b%i == 0:
        print(i, end = " ")
```

```python
#Bài 9: Nhập vào số nguyên dương a, kiểm tra xem a có phải là số chính phương hay không? Số chính phương là số có căn bậc hai là một số nguyên VD: 1, 4, 9, 16, 25....
import math
a = int(input())
can = math.sqrt(a)
if a>0:
    if can * can == a:
        print(f"{a} là số chính phương")
    else:
        print(f"{a} không là số chính phương")
else:
    print("ERROR")
```

```python
#Bài 10: Nhập vào số nguyên dương a, kiểm tra xem a có phải là số hoàn hảo hay không. Số hoàn hảo là số có tổng các ước (ước nhỏ hơn nó) bằng chính nó, VD: số 6, 28...
a = int(input())
tonguoc = 0
for i in range(1,a):
    if a%i == 0:
        tonguoc += i
if tonguoc == a and a>0:
    print(f"{a} là số hoàn hảo")
else:
    print(f"{a} không là số hoàn hảo")
```

```python
#Bài 12: Nhập số nguyên dương n, Cho S(k) = 1 + 2 + 3 + ... + k. Tìm k sao cho S(k) lớn nhất nhưng nhỏ hơn n
n = int(input())
S = 0
k = 0 
while S + (k+1) < n:
    k += 1
    S += k
print(f"Giá trị k lớn nhất là {k}, S = {S}")
```

```python
#Bài 13: Nhập vào số nguyên dương A, tìm n nhỏ nhất sao cho 1 + 1/2 + 1/3 + 1/4 + ... + 1/n > A
A = float(input())
n = 0
S = 0
while S <=A:
    n+=1
    S+= 1/n
print(f"n nhỏ nhất bằng {n}")
```

```python
#Bài 14: Nhập vào một dãy số nguyên, dừng nhập khi người dùng nhập -1. Sau khi nhập xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập. KHÔNG ĐƯỢC dùng mảng
x = int(input())
if x == -1:
    print("Chưa nhập số nào hợp lệ.")
else:
    max_val = x
    min_val = x
    while True:
        x = int(input())
        if x == -1:
            break
        if x > max_val:
            max_val = x
        if x < min_val:
            min_val = x

    print(f"Số lớn nhất: {max_val}")
    print(f"Số nhỏ nhất: {min_val}")
```

```python
#Bài 15: Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số. Làm bằng nhiều cách nhất có thể, ko dùng mảng.
n = int(input())
temp = n
count = 0
if temp == 0:
    count = 1
else:
    while temp > 0:
        count += 1
        temp //= 10
print(f"Số chữ số của {n} là: {count}")
```

```python
#Bài 16: Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số chẵn, bao nhiêu chữ số lẻ
n = int(input())
chan = 0
le = 0
temp = n
if temp ==0:
    chan = 1
else:
    while temp>0:
        chuso = temp % 10
        if chuso %2 == 0:
            chan +=1
        else:
            le +=1
        temp //= 10
print(f"n có {chan} chữ số chẵn, {le} chữ số lẻ")
    
```

```python
#Bài 18: Nhập vào một số nguyên dương n, kiểm tra xem n có phải là số dạng 3^k hay không
n = int(input())
if n <= 0:
    print(f"{n} không phải là số dạng 3^k")
else:
    temp = n
    while temp % 3 == 0:
        temp //= 3
    if temp == 1:
        print(f"{n} có dạng 3^k")
    else:
        print(f"{n} không có dạng 3^k")
```

```python
#Bài 19: Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số kế tiếp sẽ bằng tổng hai số trước đó. Nhập vào số nguyên dương A, hãy tìm số trong dãy số fibonacci lớn nhất nhưng không vượt quá A
A = int(input())
if A < 1:
    print("Không có số Fibonacci thỏa mãn.")
else:
    f1 = 1
    f2 = 1
    fk = 1 
while True:
    f3 = f1 + f2
    if f3 >A:
        break
    fk = f3
    f1 = f2
    f2 = f3
print(fk)

```

```python
#Bài 20: Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số là số nguyên tố.
n = int(input())
temp = n
cnt = 0
while temp >0:
    uoc = 0
    chuso = temp%10
    for i in range(1,chuso+1):
        if chuso % i ==0:
            uoc += 1
    if uoc ==2:
        cnt += 1
    temp //= 10
print(f"n có {cnt} chữ số nguyên tố")
```
