Bài 2: Một khu đất hình chữ nhật có chiều rộng $a$, chiều dài $b$. Người ta xây một khu vui chơi hình tròn tại vị trí trung tâm khu đất như hình vẽ. Nhập vào $a, b$. Hãy tính phần diện tích còn để trồng cây xung quanh là bao nhiêu. Biết pi = 3.14, lấy kết quả đến 2 chữ số phần thập phân.

```python
a = float(input("Nhập vào chiều rộng a:"))
b = float(input("Nhập vào chiều dài b:"))
s_hinhtron = (a/2)**2 * 3.14
print("Diện tích phần còn lại là:", round(a*b - s_hinhtron,2))
```

Bài 6: Nhập vào 3 số a, b, c. Kiểm tra xem đây có phải là cạnh của 1 tam giác không, nếu đúng thì tính diện tích của tam giác đó, in ra với 1 chữ số phần thập phân, nếu nó không phải 3 cạnh của tam giác thì in ra Khong phai 3 canh cua tam giac.

```python
import math
a = float(input("Nhập vào số đầu tiên:"))
b = float(input("Nhập vào số thứ hai:"))
c = float(input("Nhập vào số thứ ba:"))
if a>0 and b>0 and c>0 and (a+b>c) and (a+c>b) and (b+c>a):
    p= (a+b+c)/2
    S = math.sqrt(p*(p-a)*(p-b)*(p-c))
    print(f"Diện tích tam giác là {S:.1f}")
else:
    print("3 cạnh không tạo thành tam giác")
```

Bậc 3: Từ 101 - 200 kWh: 2.380 đồng/kWh
Bậc 4: Từ 201 - 300 kWh: 2.998 đồng/kWh

Bậc 6: Từ 401 trở đi kWh: 3.460 đồng/kWh
Tính thêm tiền VAT 8%. Chỉ lấy đến đơn vị đồng

```python
ten_chu_ho = input("Nhập vào họ tên chủ hộ: ")
chi_so_truoc = float(input("Nhập chỉ số điện kế tháng trước: "))
chi_so_nay = float(input("Nhập chỉ số điện kế tháng này: "))

so_kWh = chi_so_nay - chi_so_truoc

tien_dien = 0

if so_kWh > 0:

    if so_kWh <= 50:
        tien_dien = so_kWh * 1984
    elif so_kWh <= 100:
        tien_dien = 50 * 1984 + (so_kWh - 50) * 2050
    elif so_kWh <= 200:
        tien_dien = 50 * 1984 + 50 * 2050 + (so_kWh - 100) * 2380
    elif so_kWh <= 300:
        tien_dien = 50 * 1984 + 50 * 2050 + 100 * 2380 + (so_kWh - 200) * 2998
    elif so_kWh <= 400:
        tien_dien = 50 * 1984 + 50 * 2050 + 100 * 2380 + 100 * 2998 + (so_kWh - 300) * 3350
    else:
        tien_dien = 50 * 1984 + 50 * 2050 + 100 * 2380 + 100 * 2998 + 100 * 3350 + (so_kWh - 400) * 3460


    tong_tien = tien_dien * 1.08


    tong_tien_lam_tron = round(tong_tien)


    print(f"Họ và tên: {ten_chu_ho}")
    print(f"Tiền phải trả là: {tong_tien_lam_tron} đồng")
```

9. Chuyển đổi số thành chữ số: Nhập một số nguyên từ 0 đến 9, in ra tên chữ số tương ứng (0 -> "Không", 1 -> "Một", ...) - Xem trước match...case

```python
n = int(input("Nhập một số nguyên từ 0 đến 9: "))
match n:
    case 0:
        print("Không")
    case 1:
        print("Một")
    case 2:
        print("Hai")
    case 3:
        print("Ba")
    case 4:
        print("Bốn")
    case 5:
        print("Năm")
    case 6:
        print("Sáu")
    case 7:
        print("Bảy")
    case 8:
        print("Tám")
    case 9:
        print("Chín")
```
